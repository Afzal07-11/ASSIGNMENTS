```python
# ======================================
# CELL 1: Online Store Sales Analysis
# ======================================
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('sampledata.csv')

# Convert date column to datetime format
df['date'] = pd.to_datetime(df['date'])
df['month'] = df['date'].dt.month_name()
df['weekday'] = df['date'].dt.day_name()

print("--- Data Preview ---")
display(df.head())

















```

```python
# ----------------------------------------------------
# Question 1. Overall sales value & change over time
# ----------------------------------------------------
total_sales = df['total_amount'].sum()
monthly_sales = df.groupby('month', sort=False)['total_amount'].sum()

print("\n--- Q1: Overall Sales & Monthly Trend ---")
print(f"Overall Total Sales: ₹{total_sales:,.2f}")
print("\nMonthly Sales Trend:")
print(monthly_sales)
```

```python
# ----------------------------------------------------
# Question 2. Categories and Products (Largest & Smallest share)
# ----------------------------------------------------
# Category Revenue Share
category_sales = df.groupby('category')['total_amount'].sum().sort_values(ascending=False)
category_share = (category_sales / total_sales) * 100

# Product Revenue Share
product_sales = df.groupby('product_name')['total_amount'].sum().sort_values(ascending=False)

print("\n--- Q2: Category Revenue Share (%) ---")
print(pd.DataFrame({'Total Sales (₹)': category_sales, 'Share (%)': category_share.round(2)}))

print(f"\nTop Product (Largest Revenue): {product_sales.index[0]} (₹{product_sales.iloc[0]:,})")
print(f"Bottom Product (Smallest Revenue): {product_sales.index[-1]} (₹{product_sales.iloc[-1]:,})")
```

```python


# ----------------------------------------------------
# Question 3. Branch Performance (Best & Needs Investigation)
# ----------------------------------------------------
branch_performance = df.groupby('branch')['total_amount'].agg(
    total_transactions='count',
    total_sales='sum',
    average_sale='mean'
).sort_values(by='total_sales', ascending=False)

print("\n--- Q3: Branch Performance ---")
print(branch_performance)

best_branch = branch_performance.index[0]
weakest_branch = branch_performance.index[-1]
print(f"\nBest Performing Branch: {best_branch}")
print(f"Branch Needing Investigation: {weakest_branch}")
```

```python
# ----------------------------------------------------
# Question 4. Stronger vs Weaker Months & Weekdays
# ----------------------------------------------------
best_month = monthly_sales.idxmax()
worst_month = monthly_sales.idxmin()

weekday_sales = df.groupby('weekday')['total_amount'].sum().sort_values(ascending=False)
best_weekday = weekday_sales.index[0]
worst_weekday = weekday_sales.index[-1]

print("\n--- Q4: Time Period Analysis ---")
print(f"Strongest Month: {best_month} (₹{monthly_sales[best_month]:,})")
print(f"Weakest Month:   {worst_month} (₹{monthly_sales[worst_month]:,})")
print(f"Strongest Weekday: {best_weekday}")
print(f"Weaker Weekday:    {worst_weekday}")
```

```python


# ----------------------------------------------------
# Question 5. Average Transaction Value (ATV) across Categories & Locations
# ----------------------------------------------------
atv_by_category = df.groupby('category')['total_amount'].mean().round(2)
atv_by_branch = df.groupby('branch')['total_amount'].mean().round(2)

# Location vs Category Pivot Table
pivot_atv = df.pivot_table(index='branch', columns='category', values='total_amount', aggfunc='mean').round(2)

print("\n--- Q5: Average Transaction Value (ATV) ---")
print("\nATV by Category:")
print(atv_by_category)

print("\nATV by Location:")
print(atv_by_branch)

print("\nATV Matrix (Location x Category):")
print(pivot_atv)
```

```python
# ----------------------------------------------------
# Question 6. Discounts vs Quantities & Transaction Values
# ----------------------------------------------------
corr_discount_qty = df['discount_percent'].corr(df['quantity'])
corr_discount_val = df['discount_percent'].corr(df['total_amount'])

# Group by discount levels to observe patterns
discount_summary = df.groupby('discount_percent').agg(
    avg_quantity=('quantity', 'mean'),
    avg_order_value=('total_amount', 'mean')
).round(2)

print("\n--- Q6: Discount Impact Analysis ---")
print(discount_summary)
print(f"\nCorrelation between Discount % and Quantity Sold: {corr_discount_qty:.2f}")
print(f"Correlation between Discount % and Total Amount:  {corr_discount_val:.2f}")
```

```python

# ----------------------------------------------------
# Question 7. Frequent and High-Value Customers
# ----------------------------------------------------
customer_stats = df.groupby('customer_id').agg(
    order_count=('transaction_id', 'count'),
    total_spent=('total_amount', 'sum')
).sort_values(by='total_spent', ascending=False)

frequent_customers = customer_stats[customer_stats['order_count'] > 1]

print("\n--- Q7: Top Frequent & High-Value Customers ---")
print(f"Total Repeat Customers: {len(frequent_customers)}")
print("\nTop 5 Spenders:")
print(customer_stats.head(5))
```
